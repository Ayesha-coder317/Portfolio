```python
# Calculate caps safely
lower_cap = df['Transaction_Amount'].quantile(0.01)
upper_cap = df['Transaction_Amount'].quantile(0.99)

# Create capped column
df['Transaction_Amount_capped'] = df['Transaction_Amount'].clip(
    lower=lower_cap,
    upper=upper_cap
)

print("Capped column created successfully:", 'Transaction_Amount_capped' in df.columns)
```

```python
# use the capped version to avoid log(0) and extreme values
df['Transaction_Amount_log'] = np.log1p(df['Transaction_Amount_capped'])

# optional: check distribution after log
sns.histplot(df['Transaction_Amount_log'], bins=30)
plt.title("Log-transformed Transaction Amount")
plt.show()
```

It is not always a good idea to remove all outliers because some
outliers are real and meaningful. For example, in Transaction_Amount,
very large values may represent genuine high-value payments or fraud
cases. If I remove them:

- I might lose important fraud signals
- I underestimate the true risk and variability in the data

Instead of blindly removing outliers, I should understand their context
and decide whether to cap, transform or keep them.


If an extremely high Transaction_Amount is a legitimate corporate
transaction, I would not delete it. A better approach would be:

- keep the original value for audit / record
- use capping or log transformation when training models so that this
  single value does not dominate the scale
- maybe add a flag column (e.g., corporate_transaction = 1) if such
  transactions follow a specific pattern

This way I respect the real data but reduce its negative impact on
statistical modelling.


Simple IQR detection checks each column separately. If fraud happens
when both Transaction_Amount and Risk_Score are high together, each
value alone might still be within its normal range, so IQR on a single
column will miss them.

These are called compound or multivariate outliers. To detect them, I
can:

- look at scatter plots of two variables together
- apply rules that combine conditions (e.g., amount > threshold AND
  risk_score > 0.8)
- use multivariate methods such as Mahalanobis distance or clustering
  to find unusual combinations of features.

**Q: Why is handling missing values critical before modeling?**

Handling missing values is critical because most models cannot work
properly with NaNs. Missing values can:

- cause errors or force the library to silently drop rows/columns
- bias the model if the missingness is not random (e.g., high amounts
  missing more often)
- distort statistics like mean, standard deviation, and correlations

By dealing with missing values (imputation, removal, or flagging), I make
the dataset more reliable and the model more stable and interpretable.

`pd.cut()` creates bins of equal width on the numeric scale. The bin
ranges are fixed (for example, 0–200, 200–400, etc.), but the number of
rows in each bin can be very different.

`pd.qcut()` creates bins based on quantiles, so each bin has (almost) the
same number of rows, but the numeric ranges of the bins are not equal.

I would use:
- `cut` when actual value ranges matter (e.g., fixed risk thresholds).
- `qcut` when I want to compare groups of similar size, like quartiles.


Binning turns a continuous variable into categories such as "low",
"medium", and "high". This helps to:

- compare fraud rates across bins
- build simple rules (e.g., extra checks on the highest amount bin)
- make patterns easier to visualise and explain to non-technical people

For example, if the top Transaction_Amount bin has a much higher fraud
rate, that clearly signals that very large transactions are riskier.


With extreme outliers:

- `cut` keeps equal-width numeric ranges. The highest bin might contain
  only a few extreme values, while most rows sit in the middle bins.

- `qcut` adjusts the ranges so that each bin has a similar number of
  rows. The top bin may cover a very wide range because of the outliers,
  but it will still hold roughly 25% of the data (for quartiles).

For visualizing patterns and comparing bins fairly, `qcut` is usually
better because each bar represents a similar amount of data. For setting
business rules with fixed thresholds, `cut` may be more useful.
